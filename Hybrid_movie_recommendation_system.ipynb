```python
import json
```

```python
def convert3(obj):
    L=[]
    counter = 0
    for i in json.loads(obj):
        if counter<3:
            L.append(i['name'])
            counter+=1
        else:
            break
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert3)
```

```python
def fetch_director(obj):
    L=[]
    for i in json.loads(obj):
        if i['job']=='Director':
            L.append(i['name'])
            break
    return L
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
def recommend(movie,alpha=0.8):
    movie_index = movies[movies['title']==movie].index[0]
    distances = similarity[movie_index]

    recommendation = []

    for idx,sim_score in enumerate(distances):
        if idx==movie_index:
            continue
        sentiment = movies.iloc[idx]['sentiment_score']
        final_score = alpha * sim_score + (1 - alpha) * sentiment
        recommendation.append((idx,final_score))
    
    movies_list = sorted(recommendation,reverse=True,key = lambda x:x[1])[1:6]

    
    for idx,final_score in movies_list:
        print(movies.iloc[idx].title)
    
```
